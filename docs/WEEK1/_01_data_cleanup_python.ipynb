## 1 – Loading Data

We use the same dataset as in the R markdown: the **iris** dataset.

In Python we can load it directly from the `sklearn` package,  
or download the CSV from the URL used in the R version.

```python
# Option A: from sklearn (if installed)
try:
    from sklearn.datasets import load_iris
    iris_sk = load_iris(as_frame=True)
    iris = iris_sk.frame
    iris.columns = ['Sepal.Length','Sepal.Width','Petal.Length','Petal.Width','Species']
    iris['Species'] = iris['Species'].map({0:'setosa', 1:'versicolor', 2:'virginica'})
    print("Loaded via sklearn")
except ImportError:
    # Option B: read from URL  (same source as R's datasets package)
    url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
    iris = pd.read_csv(url)
    print("Loaded from URL")

print(iris.shape)   # (150, 5)
iris.head()
```

## 3 – Data Cleanup

### Handling missing values

```python
# Introduce some NaN values for demonstration
df_dirty = iris.copy()
df_dirty.loc[[5, 20, 75], 'Sepal.Length'] = np.nan
df_dirty.loc[[10, 50], 'Petal.Width'] = np.nan

print("Missing values:")
print(df_dirty.isna().sum())
```

```python
# Remove rows with any missing value  (R: na.omit)
df_clean = df_dirty.dropna()
print(f"Rows before: {len(df_dirty)}, after: {len(df_clean)}")
```

```python
# Fill missing values with column mean  (R: df$col[is.na(df$col)] <- mean(df$col, na.rm=TRUE))
df_filled = df_dirty.copy()
for col in ['Sepal.Length', 'Petal.Width']:
    df_filled[col] = df_filled[col].fillna(df_filled[col].mean())

print(df_filled.isna().sum())
```

### Filtering rows and selecting columns

```python
# Select only setosa  (R: iris[iris$Species == "setosa", ])
setosa = iris[iris['Species'] == 'setosa']
print(setosa.shape)   # (50, 5)
```

```python
# Select specific columns  (R: iris[, c("Sepal.Length", "Species")])
sub = iris[['Sepal.Length', 'Species']]
sub.head()
```

## 4 – Visualisation

### Histogram

```python
# R: hist(iris$Sepal.Length)
plt.figure(figsize=(7, 4))
plt.hist(iris['Sepal.Length'], bins=15, color='steelblue', edgecolor='white')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Count')
plt.title('Distribution of Sepal Length')
plt.tight_layout()
plt.show()
```

### Density plot

```python
# R: plot(density(iris$Sepal.Length))
plt.figure(figsize=(7, 4))
iris['Sepal.Length'].plot.kde(color='steelblue', linewidth=2)
plt.xlabel('Sepal Length (cm)')
plt.title('Density: Sepal Length')
plt.tight_layout()
plt.show()
```

### Boxplot

```python
# R: boxplot(Sepal.Length ~ Species, data=iris)
species_list = iris['Species'].unique()
data_by_species = [iris.loc[iris['Species']==sp, 'Sepal.Length'].values
                   for sp in species_list]

plt.figure(figsize=(7, 5))
plt.boxplot(data_by_species, labels=species_list, patch_artist=True)
plt.xlabel('Species')
plt.ylabel('Sepal Length (cm)')
plt.title('Sepal Length by Species')
plt.tight_layout()
plt.show()
```

### QQ-plot  *(check for normality)*

```python
# R: qqnorm(iris$Sepal.Length); qqline(iris$Sepal.Length)
fig, ax = plt.subplots(figsize=(5, 5))
stats.probplot(iris['Sepal.Length'], dist='norm', plot=ax)
ax.set_title('QQ-Plot: Sepal Length')
plt.tight_layout()
plt.show()
```
