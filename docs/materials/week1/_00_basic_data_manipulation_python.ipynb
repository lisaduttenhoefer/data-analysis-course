## 1 – Loops

### apply / sapply / lapply  →  `np.mean(axis=)` and list comprehensions

```python
Mat = np.random.randn(3, 4)
print("Matrix:")
print(Mat)

# apply(Mat, 1, mean) — mean of each ROW
print("Row means:   ", np.mean(Mat, axis=1))

# apply(Mat, 2, mean) — mean of each COLUMN
print("Column means:", np.mean(Mat, axis=0))
```

```python
# sapply(x, sqrt)  →  list comprehension or map()
x = list(range(1, 6))          # [1, 2, 3, 4, 5]

# list comprehension (most Pythonic)
result = [np.sqrt(i) for i in x]
print(result)

# map() — equivalent to sapply, returns an iterator
result2 = list(map(np.sqrt, x))
print(result2)
```

### for loops

```python
# R: for (i in 1:4) { print(i * 3) }
for i in range(1, 5):     # range(1,5) = 1, 2, 3, 4
    print(i * 3)
```

```python
# Looping over a list directly
colors = ["blue", "red", "green"]
for color in colors:
    print(f"Current color: {color}")
```

```python
# enumerate() — get index AND value at the same time
for i, color in enumerate(colors):
    print(f"Index {i}: {color}")
```

```python
# Accumulating results in a loop
total = 0
for i in range(1, 11):
    total += i
print("Sum 1-10:", total)  # 55
```

## 2 – If / Else Statements

```python
a = 5
b = 10

if a < b:
    print("a is smaller")
else:
    print("b is smaller or equal")
```

```python
# elif — multiple conditions  (R: else if)
score = 72

if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "F"

print(f"Score {score} → Grade {grade}")
```

```python
# Combine loops and if statements
numbers = [4, 7, 2, 9, 1, 6]

for n in numbers:
    if n > 5:
        print(f"{n} is large")
    else:
        print(f"{n} is small")
```

## 3 – Index Selection

### `np.where`  *(equivalent to R: `which`)*

```python
x = np.array([2, 8, 3, 7, 1])

# R: which(x > 5)
idx = np.where(x > 5)
print("Indices where x > 5:", idx[0])   # [1 3]

# R: x[which(x > 5)]
print("Values where x > 5: ", x[idx])   # [8 7]
```

```python
# Direct boolean indexing — shortcut  (R: x[x > 5])
print(x[x > 5])    # [8 7]
```

### `in`  *(equivalent to R: `%in%`)*

```python
# R: "a" %in% c("b", "d")
print("a" in ["b", "d"])   # False
print("b" in ["b", "d"])   # True
```

```python
# Check each element of one list against another  (R: c("a","b","c") %in% c("b","d"))
source = ["a", "b", "c"]
target = ["b", "d"]

result = [s in target for s in source]
print(result)   # [False, True, False]

# With NumPy
print(np.isin(source, target))   # [False  True False]
```

### `pd.isna` / `np.isnan`  *(equivalent to R: `is.na`)*

```python
s = pd.Series([1, None, 3, float('nan'), 5])

# R: is.na(s)
print(pd.isna(s))
print("Count of missing:", pd.isna(s).sum())
```

```python
# Filter out missing values  (R: na.omit / s[!is.na(s)])
clean = s.dropna()
print(clean)
```
