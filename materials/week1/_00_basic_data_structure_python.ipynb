## 2 – First steps: simple variables

In Python, `=` is used for assignment (no `<-` like in R).  
Python is **case-sensitive**: `hello` and `Hello` are different variables.

```python
a    = 2          # integer
b    = 3.14       # float
c    = True       # boolean (capital T!)
name = "Rosalinde"  # string

print(a, b, c, name)
print(type(a), type(b), type(c), type(name))
```

```python
# Arithmetic with variables
x = 5
y = 3
total = x + y
print(total * 5)
```

## 3 – Lists  *(equivalent to R vectors)*

Python lists use **0-based indexing** (R uses 1-based).  
Lists can hold mixed types, unlike R numeric vectors.

```python
y = [9, 3, 7, 2]
print(len(y))      # 4

# Indexing — 0-based!
print(y[0])        # 9  (R: y[1])
print(y[2])        # 7  (R: y[3])
print(y[1:3])      # [3, 7]  — index 1 and 2 (R: y[2:3])
print(y[-1])       # 2  — last element
```

```python
# Filtering with list comprehension  (R: y[y < 5])
small = [x for x in y if x < 5]
print(small)  # [3, 2]

# Count elements that match a condition  (R: sum(y < 5))
print(sum(x < 5 for x in y))  # 2
```

```python
# Modifying a list
y[0] = 10     # replace first element  (R: y[1] <- 10)
y.append(99)  # add element at the end  (no R equivalent — use c(y, 99))
print(y)
```

## 4 – NumPy Arrays  *(equivalent to R matrices)*

NumPy arrays support multi-dimensional data and fast math operations.  
Indexing is **0-based** and uses `[row, column]`.

```python
X = np.array([[1, 2, 3],
              [4, 5, 6]])
print(X)
print("Shape:", X.shape)  # (2, 3) — rows, columns
```

```python
# Indexing
print(X[0, 1])   # 2   — row 0, column 1  (R: X[1,2])
print(X[1, :])   # [4 5 6] — entire second row  (R: X[2,])
print(X[:, 0])   # [1 4]   — entire first column  (R: X[,1])
```

```python
# Building arrays from vectors
A = np.column_stack(([1,2], [3,4], [5,6]))   # like R: cbind(c(1,2), c(3,4), c(5,6))
B = np.vstack([[1,2,3], [4,5,6]])             # like R: rbind(c(1,2,3), c(4,5,6))
print(A)
print(B)
```

## 5 – Dictionaries  *(equivalent to R lists)*

Python dicts hold key–value pairs.  
Access with `d["key"]` — no `$` or `[[]]` like in R.

```python
L = {"color": ["blue", "red"], "size": 5, "active": True}

print(L["color"])       # ['blue', 'red']  (R: L[["color"]])
print(L["color"][1])    # 'red'  — 0-based index  (R: L[["color"]][2])
print("size" in L)      # True  — check if key exists
```

```python
# Modify and delete
L["size"] = 10      # change a value
L["new"] = "hello" # add a new key
del L["new"]        # delete a key  (R: L[["new"]] <- NULL)
print(L)
```

## 6 – Pandas DataFrames  *(equivalent to R data.frame)*

DataFrames are the main structure for tabular data.  
Each column can have a different type.

```python
name   = ["Ben", "Alice", "Daniel", "Charlie"]
year   = [1990, 1992, 1995, 1993]
option = ["Biology", "Physics", "Chemistry", "Mathematics"]

df = pd.DataFrame({"First_Name": name, "Year": year, "Options": option})
print(df)
```

```python
# Accessing columns
print(df["Year"])          # (R: df$Year)
print(df.iloc[:, 1])       # same — column index 1  (R: df[,2])
```

```python
# Accessing single elements
print(df["Options"].iloc[2])    # 'Chemistry'  (R: df$Options[3])
print(df.iloc[1:3, 2])          # rows 1-2, column 2  (R: df[2:3, 3])
```

```python
# Merge two DataFrames  (R: merge)
full_names = pd.DataFrame({"First_Name": ["Alice", "Ben"],
                            "Last_Name":  ["Parker", "James"]})

inner = pd.merge(df, full_names)               # inner join
outer = pd.merge(df, full_names, how="outer")  # outer join — NaN for missing
print(inner)
print(outer)
```

## Summary: What have we learned?

| R | Python | Description |
|---|--------|-------------|
| `a <- 5` | `a = 5` | Variable assignment |
| `c(1, 2, 3)` | `[1, 2, 3]` | Ordered sequence |
| `matrix(...)` | `np.array([[...]])` | 2-D numerical data |
| `list(a=1, b=2)` | `{"a": 1, "b": 2}` | Named key–value pairs |
| `data.frame(...)` | `pd.DataFrame({...})` | Tabular data |
| `x[2]` | `x[1]` | Second element (0-based vs 1-based) |
| `merge(a, b)` | `pd.merge(a, b)` | Join two tables |
